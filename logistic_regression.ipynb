```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix # evaluate the model
```

```python
df = pd.read_csv('data/womens_clothing_ecommerce_reviews.csv')
```

```python
df.head()
```

```python
print(f'Null value count : {df.isnull().sum()}')
print(f'Duplicate values : {df.duplicated().sum()}')
print(f"Empty review count : {int((df['Review Text'].fillna('').str.strip() == '').sum())}")
```

```python
print('Class counts (imbalance matters for accuracy) : ')
print(df['sentiment'].value_counts())
print('\nClass share : ')
print(df['sentiment'].value_counts(normalize=True).round(4))
print(f"\nLabel values : {sorted(df['sentiment'].unique().tolist())}")
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
x = df['Review Text']
y = df['sentiment']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)

print(f'x shape : {x.shape}')
print(f'y shape : {y.shape}')
print(f'x_train : {x_train.shape}')
print(f'y_train : {y_train.shape}')
print(f'x_test : {x_test.shape}')
print(f'y_test : {y_test.shape}')
```

```python
print("Converting text to numerical features using Bag-of-Words technique")
vectorizer = CountVectorizer(stop_words='english')

x_train_bow = vectorizer.fit_transform(x_train)

print(f'x_train_bow : {x_train_bow.shape}')
print(f'Unique words (features) : {vectorizer.get_feature_names_out()}')
```

```python
print(f"x_train_bow : {x_train_bow}")
```

```python
x_test_bow = vectorizer.transform(x_test)
```

```python
model = LogisticRegression(max_iter=1000, solver='lbfgs', random_state=42)

model.fit(x_train_bow, y_train)

print("👌 model training completed.")
```

```python
y_test_prep = model.predict(x_test_bow)
y_train_prep = model.predict(x_train_bow)

# Calculate train accuracy
accuracy = accuracy_score(y_train, y_train_prep)
print(f'Accuracy (Train) : {accuracy:.4f} ({accuracy:.2%})')

# Calculate test accuracy
accuracy = accuracy_score(y_test, y_test_prep)
print(f'Accuracy (Test) : {accuracy:.4f} ({accuracy:.2%})')
```
